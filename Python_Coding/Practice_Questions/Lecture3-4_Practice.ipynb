## 2. Combining Data Structures

You are given a list of tuples, where each tuple contains a name (string) and an age (integer). Write a function that returns a dictionary mapping each age to a set of names of people who have that age.

Hints:<br>
- Iterate over the list of tuples.
- For each tuple, use the age as a key in the dictionary.
- If that age already exists, add the name to the set; otherwise, create a new set.

This demonstrates the use of both dictionaries and sets.

Example:

Input: 
```Python
[('Alice', 30), ('Bob', 25), ('Charlie', 30), ('David', 25), ('Eve', 35)]
```
Output:
```Python
{
  30: {'Alice', 'Charlie'},
  25: {'Bob', 'David'},
  35: {'Eve'}
}
```

## 6. Factorial Function using Recursion

Write a recursive function named factorial that accepts a positive integer and returns its factorial (n!).

Remember that the factorial of 0 is 1, and the factorial of any positive integer n is n * factorial(n-1).

Hints:
- Use a base case to handle when n is 0.
- Call the function recursively with (n-1).

Example:
```Python
Input: factorial(5)
Expected Output: 120
```
(The factorial of 5 is 5 * 4 * 3 * 2 * 1 = 120)
